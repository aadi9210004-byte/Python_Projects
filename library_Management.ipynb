```python

from abc import ABC, abstractmethod
import csv
import os


# ==========================================================
# ABSTRACT CLASS
# ==========================================================

class Record(ABC):

    @abstractmethod
    def display(self):
        pass


# ==========================================================
# BOOK CLASS
# ==========================================================

class Book(Record):

    def __init__(self, book_id, title, author, category,
                 total_copies, available_copies=None):

        self.__book_id = book_id
        self.__title = title
        self.__author = author
        self.__category = category
        self.__total_copies = total_copies

        if available_copies is None:
            available_copies = total_copies

        self.__available_copies = available_copies
        self.__borrowers = []

    # Properties

    @property
    def book_id(self):
        return self.__book_id

    @property
    def title(self):
        return self.__title

    @property
    def author(self):
        return self.__author

    @property
    def category(self):
        return self.__category

    @property
    def total_copies(self):
        return self.__total_copies

    @property
    def available_copies(self):
        return self.__available_copies

    @property
    def borrowers(self):
        return self.__borrowers.copy()

    # Issue a book

    def issue(self, student_name):

        if self.__available_copies <= 0:
            return False

        self.__available_copies -= 1
        self.__borrowers.append(student_name)

        return True

    # Return a book

    def return_book(self, student_name):

        if student_name not in self.__borrowers:
            return False

        self.__borrowers.remove(student_name)
        self.__available_copies += 1

        return True

    # Display book information

    def display(self):

        return [
            self.book_id,
            self.title,
            self.author,
            self.category,
            self.total_copies,
            self.available_copies
        ]


# ==========================================================
# LIBRARY MANAGEMENT CLASS
# ==========================================================

class LibraryManagementSystem:

    def __init__(self):

        self.books = []

    # ------------------------------------------------------
    # ADD BOOK
    # ------------------------------------------------------

    def add_book(self, book):

        for existing_book in self.books:

            if existing_book.book_id == book.book_id:
                print("Error: Book ID already exists.")
                return

        self.books.append(book)
        print("Book added successfully!")

    # ------------------------------------------------------
    # FIND BOOK BY ID
    # ------------------------------------------------------

    def find_book(self, book_id):

        for book in self.books:

            if book.book_id == book_id:
                return book

        return None

    # ------------------------------------------------------
    # SEARCH BOOK
    # ------------------------------------------------------

    def search_book(self, search_term):

        matches = []

        for book in self.books:

            if (
                search_term.lower() in book.title.lower()
                or search_term == book.book_id
            ):
                matches.append(book)

        if not matches:
            print("No matching books found.")
            return

        print("\nSearch Results")
        print("-" * 60)

        for book in matches:
            print("Book ID:", book.book_id)
            print("Title:", book.title)
            print("Author:", book.author)
            print("Category:", book.category)
            print("Available copies:", book.available_copies)
            print("-" * 60)

    # ------------------------------------------------------
    # DISPLAY BOOKS
    # ------------------------------------------------------

    def display_books(self, available_only=False):

        selected_books = []

        for book in self.books:

            if not available_only or book.available_copies > 0:
                selected_books.append(book)

        if not selected_books:
            print("No books to display.")
            return

        print("\n" + "=" * 105)

        print(
            f"{'ID':<10}"
            f"{'Title':<25}"
            f"{'Author':<20}"
            f"{'Category':<18}"
            f"{'Total':<10}"
            f"{'Available':<10}"
        )

        print("=" * 105)

        for book in selected_books:

            print(
                f"{book.book_id:<10}"
                f"{book.title[:24]:<25}"
                f"{book.author[:19]:<20}"
                f"{book.category[:17]:<18}"
                f"{book.total_copies:<10}"
                f"{book.available_copies:<10}"
            )

        print("=" * 105)

    # ------------------------------------------------------
    # ISSUE BOOK
    # ------------------------------------------------------

    def issue_book(self, book_id, student_name):

        book = self.find_book(book_id)

        if book is None:
            print("Book not found.")
            return

        if book.issue(student_name):
            print(
                f'"{book.title}" issued successfully '
                f"to {student_name}."
            )
        else:
            print("No copies of this book are available.")

    # ------------------------------------------------------
    # RETURN BOOK
    # ------------------------------------------------------

    def return_book(self, book_id, student_name):

        book = self.find_book(book_id)

        if book is None:
            print("Book not found.")
            return

        if book.return_book(student_name):
            print(
                f'"{book.title}" returned successfully '
                f"by {student_name}."
            )
        else:
            print(
                "Return unsuccessful. Check the book ID "
                "and borrower name."
            )

    # ------------------------------------------------------
    # SAVE RECORDS TO CSV
    # ------------------------------------------------------

    def save_csv(self):

        try:
            with open(
                "books.csv", "w", newline="", encoding="utf-8"
            ) as file:

                writer = csv.writer(file)

                writer.writerow([
                    "Book ID",
                    "Title",
                    "Author",
                    "Category",
                    "Total Copies",
                    "Available Copies",
                    "Borrowers"
                ])

                for book in self.books:

                    writer.writerow([
                        book.book_id,
                        book.title,
                        book.author,
                        book.category,
                        book.total_copies,
                        book.available_copies,
                        "|".join(book.borrowers)
                    ])

            print("Book records saved to books.csv.")

        except OSError as error:
            print("Error saving CSV:", error)

    # ------------------------------------------------------
    # SAVE RECORDS TO TXT
    # ------------------------------------------------------

    def save_txt(self):

        try:
            with open(
                "books.txt", "w", encoding="utf-8"
            ) as file:

                file.write("LIBRARY MANAGEMENT SYSTEM\n")
                file.write("=" * 60 + "\n\n")

                for book in self.books:

                    file.write(f"Book ID: {book.book_id}\n")
                    file.write(f"Title: {book.title}\n")
                    file.write(f"Author: {book.author}\n")
                    file.write(f"Category: {book.category}\n")
                    file.write(
                        f"Total copies: {book.total_copies}\n"
                    )
                    file.write(
                        f"Available copies: "
                        f"{book.available_copies}\n"
                    )

                    borrowers = book.borrowers

                    file.write(
                        "Borrowers: "
                        + (", ".join(borrowers) if borrowers else "None")
                        + "\n"
                    )

                    file.write("-" * 60 + "\n")

            print("Book records saved to books.txt.")

        except OSError as error:
            print("Error saving TXT:", error)

    # ------------------------------------------------------
    # LOAD RECORDS FROM CSV
    # ------------------------------------------------------

    def load_csv(self):

        if not os.path.exists("books.csv"):
            print("No existing book records found.")
            return

        try:
            loaded_books = []

            with open(
                "books.csv", "r", newline="", encoding="utf-8"
            ) as file:

                reader = csv.DictReader(file)

                for row in reader:

                    total = int(row["Total Copies"])
                    available = int(row["Available Copies"])

                    if (
                        total < 0
                        or available < 0
                        or available > total
                    ):
                        raise ValueError("Invalid copy counts in CSV.")

                    book = Book(
                        row["Book ID"],
                        row["Title"],
                        row["Author"],
                        row["Category"],
                        total,
                        available
                    )

                    borrowers_text = row.get("Borrowers", "")
                    borrowers = (
                        borrowers_text.split("|")
                        if borrowers_text else []
                    )

                    if len(borrowers) != total - available:
                        raise ValueError(
                            f"Borrower count does not match "
                            f"issued copies for {book.title}."
                        )

                    for borrower in borrowers:
                        if not borrower:
                            raise ValueError("Empty borrower name.")

                        book.issue(borrower)

                    if any(
                        existing.book_id == book.book_id
                        for existing in loaded_books
                    ):
                        raise ValueError("Duplicate book ID in CSV.")

                    loaded_books.append(book)

            self.books = loaded_books
            print("Book records loaded successfully.")

        except (OSError, ValueError, KeyError) as error:
            print("Could not load book records:", error)


# ==========================================================
# INPUT HELPER FUNCTIONS
# ==========================================================

def get_nonempty(prompt):

    while True:

        value = input(prompt).strip()

        if value:
            return value

        print("This field cannot be empty.")


def get_integer(prompt, minimum=0):

    while True:

        try:
            value = int(input(prompt))

            if value < minimum:
                print(f"Enter a number of at least {minimum}.")
                continue

            return value

        except ValueError:
            print("Please enter a valid whole number.")


# ==========================================================
# MAIN PROGRAM
# ==========================================================

library = LibraryManagementSystem()

# Load previously saved records
library.load_csv()


while True:

    print("\n" + "=" * 45)
    print("       LIBRARY MANAGEMENT SYSTEM")
    print("=" * 45)

    print("1. Add Book")
    print("2. Search Books")
    print("3. Issue Book")
    print("4. Return Book")
    print("5. Display Available Books")
    print("6. Display All Books")
    print("7. Save Records to CSV")
    print("8. Save Records to TXT")
    print("9. Save Records to Both Files")
    print("10. Exit")

    print("=" * 45)

    choice = input("Enter your choice: ").strip()

    try:

        # ADD BOOK
        if choice == "1":

            print("\n--- Add Book ---")

            book_id = get_nonempty("Book ID: ")
            title = get_nonempty("Book title: ")
            author = get_nonempty("Author: ")
            category = get_nonempty("Category: ")

            total_copies = get_integer(
                "Number of copies: ", minimum=1
            )

            book = Book(
                book_id,
                title,
                author,
                category,
                total_copies
            )

            library.add_book(book)

        # SEARCH BOOK
        elif choice == "2":

            term = get_nonempty(
                "Enter book ID or title: "
            )

            library.search_book(term)

        # ISSUE BOOK
        elif choice == "3":

            book_id = get_nonempty("Enter book ID: ")
            student_name = get_nonempty(
                "Enter borrower's name: "
            )

            library.issue_book(book_id, student_name)

        # RETURN BOOK
        elif choice == "4":

            book_id = get_nonempty("Enter book ID: ")
            student_name = get_nonempty(
                "Enter borrower's name: "
            )

            library.return_book(book_id, student_name)

        # DISPLAY AVAILABLE BOOKS
        elif choice == "5":

            library.display_books(available_only=True)

        # DISPLAY ALL BOOKS
        elif choice == "6":

            library.display_books()

        # SAVE CSV
        elif choice == "7":

            library.save_csv()

        # SAVE TXT
        elif choice == "8":

            library.save_txt()

        # SAVE BOTH FILES
        elif choice == "9":

            library.save_csv()
            library.save_txt()

        # EXIT
        elif choice == "10":

            library.save_csv()
            library.save_txt()

            print("Records saved. Goodbye!")
            break

        else:

            print("Invalid choice. Enter a number from 1 to 10.")

    except (ValueError, OSError) as error:

        print("Operation failed:", error)
```
