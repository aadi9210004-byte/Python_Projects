```python

from abc import ABC, abstractmethod
import csv


# ============================================================
# ABSTRACT CLASS
# ============================================================

class Record(ABC):

    @abstractmethod
    def display(self):
        pass


# ============================================================
# STUDENT CLASS
# ============================================================

class Student(Record):

    def __init__(self, student_id, name, age, course,
                 math, python, data_structures):

        self.__student_id = student_id
        self.__name = name
        self.__age = age
        self.__course = course
        self.__math = math
        self.__python = python
        self.__data_structures = data_structures

    # --------------------------------------------------------
    # Properties
    # --------------------------------------------------------

    @property
    def student_id(self):
        return self.__student_id

    @property
    def name(self):
        return self.__name

    @property
    def age(self):
        return self.__age

    @property
    def course(self):
        return self.__course

    @property
    def math(self):
        return self.__math

    @property
    def python(self):
        return self.__python

    @property
    def data_structures(self):
        return self.__data_structures

    # --------------------------------------------------------
    # Average marks
    # --------------------------------------------------------

    def average_marks(self):

        return (
            self.__math +
            self.__python +
            self.__data_structures
        ) / 3

    # --------------------------------------------------------
    # Display student
    # --------------------------------------------------------

    def display(self):

        return [
            self.student_id,
            self.name,
            self.age,
            self.course,
            self.math,
            self.python,
            self.data_structures,
            round(self.average_marks(), 2)
        ]


# ============================================================
# STUDENT MANAGEMENT SYSTEM
# ============================================================

class StudentManagementSystem:

    def __init__(self):

        self.students = []

    # --------------------------------------------------------
    # ADD STUDENT
    # --------------------------------------------------------

    def add_student(self, student):

        # Check duplicate ID

        for s in self.students:

            if s.student_id == student.student_id:

                print("Student ID already exists.")
                return

        self.students.append(student)

        print("Student added successfully.")

    # --------------------------------------------------------
    # DISPLAY STUDENTS
    # --------------------------------------------------------

    def display_students(self):

        if not self.students:

            print("\nNo student records found.")
            return

        print("\n" + "=" * 100)

        print(
            f"{'ID':<10}"
            f"{'Name':<20}"
            f"{'Age':<8}"
            f"{'Course':<20}"
            f"{'Math':<8}"
            f"{'Python':<10}"
            f"{'DS':<8}"
            f"{'Average':<10}"
        )

        print("=" * 100)

        for student in self.students:

            print(
                f"{student.student_id:<10}"
                f"{student.name:<20}"
                f"{student.age:<8}"
                f"{student.course:<20}"
                f"{student.math:<8}"
                f"{student.python:<10}"
                f"{student.data_structures:<8}"
                f"{student.average_marks():<10.2f}"
            )

        print("=" * 100)

    # --------------------------------------------------------
    # SEARCH STUDENT
    # --------------------------------------------------------

    def search_student(self, student_id):

        for student in self.students:

            if student.student_id == student_id:

                print("\nStudent Found")
                print("-" * 40)

                print("Student ID :", student.student_id)
                print("Name       :", student.name)
                print("Age        :", student.age)
                print("Course     :", student.course)
                print("Math       :", student.math)
                print("Python     :", student.python)
                print("Data Struct:", student.data_structures)
                print("Average    :", round(student.average_marks(), 2))

                return

        print("Student not found.")

    # --------------------------------------------------------
    # UPDATE STUDENT
    # --------------------------------------------------------

    def update_student(self, student_id):

        for index, student in enumerate(self.students):

            if student.student_id == student_id:

                print("\nCurrent Student Details")

                print("Name       :", student.name)
                print("Age        :", student.age)
                print("Course     :", student.course)
                print("Math       :", student.math)
                print("Python     :", student.python)
                print("Data Struct:", student.data_structures)

                print("\nEnter new details")

                name = input("Name: ")
                age = int(input("Age: "))
                course = input("Course: ")

                math = float(input("Math marks: "))
                python = float(input("Python marks: "))
                data_structures = float(
                    input("Data Structures marks: ")
                )

                updated_student = Student(
                    student_id,
                    name,
                    age,
                    course,
                    math,
                    python,
                    data_structures
                )

                self.students[index] = updated_student

                print("Student record updated successfully.")

                return

        print("Student not found.")

    # --------------------------------------------------------
    # DELETE STUDENT
    # --------------------------------------------------------

    def delete_student(self, student_id):

        for student in self.students:

            if student.student_id == student_id:

                self.students.remove(student)

                print("Student deleted successfully.")

                return

        print("Student not found.")

    # --------------------------------------------------------
    # CALCULATE AVERAGE MARKS
    # --------------------------------------------------------

    def calculate_average(self):

        if not self.students:

            print("No student records available.")
            return

        total = 0

        for student in self.students:

            total += student.average_marks()

        average = total / len(self.students)

        print("\nOverall Class Average:", round(average, 2))

    # --------------------------------------------------------
    # SAVE TO CSV
    # --------------------------------------------------------

    def save_csv(self):

        try:

            with open(
                "students.csv",
                "w",
                newline=""
            ) as file:

                writer = csv.writer(file)

                # Header

                writer.writerow([
                    "Student ID",
                    "Name",
                    "Age",
                    "Course",
                    "Math",
                    "Python",
                    "Data Structures",
                    "Average"
                ])

                # Student records

                for student in self.students:

                    writer.writerow(
                        student.display()
                    )

            print("Records saved to students.csv")

        except Exception as e:

            print("Error saving CSV file:", e)

    # --------------------------------------------------------
    # SAVE TO TXT
    # --------------------------------------------------------

    def save_txt(self):

        try:

            with open(
                "students.txt",
                "w"
            ) as file:

                file.write(
                    "STUDENT MANAGEMENT SYSTEM\n"
                )

                file.write(
                    "=" * 80 + "\n"
                )

                for student in self.students:

                    file.write(
                        f"Student ID: {student.student_id}\n"
                    )

                    file.write(
                        f"Name: {student.name}\n"
                    )

                    file.write(
                        f"Age: {student.age}\n"
                    )

                    file.write(
                        f"Course: {student.course}\n"
                    )

                    file.write(
                        f"Math: {student.math}\n"
                    )

                    file.write(
                        f"Python: {student.python}\n"
                    )

                    file.write(
                        f"Data Structures: "
                        f"{student.data_structures}\n"
                    )

                    file.write(
                        f"Average: "
                        f"{student.average_marks():.2f}\n"
                    )

                    file.write(
                        "-" * 80 + "\n"
                    )

            print("Records saved to students.txt")

        except Exception as e:

            print("Error saving TXT file:", e)

    # --------------------------------------------------------
    # LOAD FROM CSV
    # --------------------------------------------------------

    def load_csv(self):

        try:

            with open(
                "students.csv",
                "r"
            ) as file:

                reader = csv.DictReader(file)

                self.students.clear()

                for row in reader:

                    student = Student(
                        row["Student ID"],
                        row["Name"],
                        int(row["Age"]),
                        row["Course"],
                        float(row["Math"]),
                        float(row["Python"]),
                        float(row["Data Structures"])
                    )

                    self.students.append(student)

            print("Records loaded from students.csv")

        except FileNotFoundError:

            print(
                "No existing students.csv file found."
            )

        except Exception as e:

            print("Error loading CSV:", e)


# ============================================================
# INPUT FUNCTIONS
# ============================================================

def create_student():

    print("\n===== ADD STUDENT =====")

    student_id = input(
        "Student ID: "
    )

    name = input(
        "Name: "
    )

    age = int(
        input("Age: ")
    )

    course = input(
        "Course: "
    )

    math = float(
        input("Math marks: ")
    )

    python = float(
        input("Python marks: ")
    )

    data_structures = float(
        input("Data Structures marks: ")
    )

    return Student(
        student_id,
        name,
        age,
        course,
        math,
        python,
        data_structures
    )


# ============================================================
# MAIN PROGRAM
# ============================================================

system = StudentManagementSystem()

# Automatically load previous records

system.load_csv()


while True:

    print("\n")
    print("=" * 50)
    print("       STUDENT MANAGEMENT SYSTEM")
    print("=" * 50)

    print("1. Add Student")
    print("2. Display Students")
    print("3. Search Student")
    print("4. Update Student")
    print("5. Delete Student")
    print("6. Calculate Average Marks")
    print("7. Save Records to CSV")
    print("8. Save Records to TXT")
    print("9. Save to Both CSV and TXT")
    print("10. Exit")

    print("=" * 50)

    try:

        choice = int(
            input("Enter your choice: ")
        )

        # ----------------------------------------------------
        # ADD
        # ----------------------------------------------------

        if choice == 1:

            student = create_student()

            system.add_student(student)

        # ----------------------------------------------------
        # DISPLAY
        # ----------------------------------------------------

        elif choice == 2:

            system.display_students()

        # ----------------------------------------------------
        # SEARCH
        # ----------------------------------------------------

        elif choice == 3:

            student_id = input(
                "Enter Student ID to search: "
            )

            system.search_student(student_id)

        # ----------------------------------------------------
        # UPDATE
        # ----------------------------------------------------

        elif choice == 4:

            student_id = input(
                "Enter Student ID to update: "
            )

            system.update_student(student_id)

        # ----------------------------------------------------
        # DELETE
        # ----------------------------------------------------

        elif choice == 5:

            student_id = input(
                "Enter Student ID to delete: "
            )

            system.delete_student(student_id)

        # ----------------------------------------------------
        # AVERAGE
        # ----------------------------------------------------

        elif choice == 6:

            system.calculate_average()

        # ----------------------------------------------------
        # SAVE CSV
        # ----------------------------------------------------

        elif choice == 7:

            system.save_csv()

        # ----------------------------------------------------
        # SAVE TXT
        # ----------------------------------------------------

        elif choice == 8:

            system.save_txt()

        # ----------------------------------------------------
        # SAVE BOTH
        # ----------------------------------------------------

        elif choice == 9:

            system.save_csv()
            system.save_txt()

        # ----------------------------------------------------
        # EXIT
        # ----------------------------------------------------

        elif choice == 10:

            # Automatically save before exiting

            system.save_csv()
            system.save_txt()

            print("\nThank you for using")
            print("Student Management System.")

            break

        else:

            print(
                "Invalid choice. "
                "Please enter a number from 1-10."
            )

    except ValueError:

        print(
            "\nInvalid input."
            "\nPlease enter the correct data type."
        )

    except Exception as e:

        print(
            "\nAn unexpected error occurred:",
            e
        )
```
