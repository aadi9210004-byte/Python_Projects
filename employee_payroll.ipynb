```python

from abc import ABC, abstractmethod
from decimal import Decimal, InvalidOperation
from datetime import datetime
import csv
import os


# ==========================================================
# ABSTRACT CLASS
# ==========================================================

class Record(ABC):

    @abstractmethod
    def display(self):
        pass


# ==========================================================
# EMPLOYEE CLASS
# ==========================================================

class Employee(Record):

    def __init__(
        self, employee_id, name, designation,
        basic_salary, allowances, deductions
    ):

        self.__employee_id = employee_id
        self.__name = name
        self.__designation = designation
        self.__basic_salary = Decimal(basic_salary)
        self.__allowances = Decimal(allowances)
        self.__deductions = Decimal(deductions)

    # Properties

    @property
    def employee_id(self):
        return self.__employee_id

    @property
    def name(self):
        return self.__name

    @property
    def designation(self):
        return self.__designation

    @property
    def basic_salary(self):
        return self.__basic_salary

    @property
    def allowances(self):
        return self.__allowances

    @property
    def deductions(self):
        return self.__deductions

    # Calculate gross salary

    def gross_salary(self):

        return self.__basic_salary + self.__allowances

    # Calculate net salary

    def net_salary(self):

        return self.gross_salary() - self.__deductions

    # Display employee details

    def display(self):

        print("-" * 55)
        print("Employee ID :", self.employee_id)
        print("Name        :", self.name)
        print("Designation :", self.designation)
        print("Basic Salary: Rs.", self.basic_salary)
        print("Allowances  : Rs.", self.allowances)
        print("Deductions  : Rs.", self.deductions)
        print("Gross Salary: Rs.", self.gross_salary())
        print("Net Salary  : Rs.", self.net_salary())
        print("-" * 55)


# ==========================================================
# EMPLOYEE PAYROLL MANAGEMENT SYSTEM
# ==========================================================

class EmployeePayrollSystem:

    def __init__(self):

        self.employees = []
        self.load_csv()

    # ------------------------------------------------------
    # FIND EMPLOYEE
    # ------------------------------------------------------

    def find_employee(self, employee_id):

        for employee in self.employees:

            if employee.employee_id == employee_id:
                return employee

        return None

    # ------------------------------------------------------
    # ADD EMPLOYEE
    # ------------------------------------------------------

    def add_employee(self, employee):

        if self.find_employee(employee.employee_id):

            print("Error: Employee ID already exists.")
            return

        self.employees.append(employee)

        print("Employee added successfully!")

        self.save_csv()

    # ------------------------------------------------------
    # SEARCH EMPLOYEE
    # ------------------------------------------------------

    def search_employee(self, employee_id):

        employee = self.find_employee(employee_id)

        if employee is None:

            print("Employee not found.")
            return

        print("\nEmployee Found")
        employee.display()

    # ------------------------------------------------------
    # DISPLAY ALL EMPLOYEES
    # ------------------------------------------------------

    def display_employees(self):

        if not self.employees:

            print("No employee records found.")
            return

        print("\nALL EMPLOYEES")
        print("=" * 95)

        print(
            f"{'ID':<12}"
            f"{'Name':<22}"
            f"{'Designation':<20}"
            f"{'Basic Salary':<15}"
            f"{'Net Salary':<15}"
        )

        print("=" * 95)

        for employee in self.employees:

            print(
                f"{employee.employee_id:<12}"
                f"{employee.name[:21]:<22}"
                f"{employee.designation[:19]:<20}"
                f"{employee.basic_salary:<15.2f}"
                f"{employee.net_salary():<15.2f}"
            )

        print("=" * 95)

    # ------------------------------------------------------
    # CALCULATE SALARY
    # ------------------------------------------------------

    def calculate_salary(self, employee_id):

        employee = self.find_employee(employee_id)

        if employee is None:

            print("Employee not found.")
            return

        print("\nSALARY CALCULATION")

        print("Employee Name :", employee.name)
        print("Basic Salary  : Rs.", employee.basic_salary)
        print("Allowances    : Rs.", employee.allowances)
        print("Gross Salary  : Rs.", employee.gross_salary())
        print("Deductions    : Rs.", employee.deductions)
        print("Net Salary    : Rs.", employee.net_salary())

    # ------------------------------------------------------
    # GENERATE PAYSLIP
    # ------------------------------------------------------

    def generate_payslip(self, employee_id):

        employee = self.find_employee(employee_id)

        if employee is None:

            print("Employee not found.")
            return

        filename = f"payslip_{employee.employee_id}.txt"

        try:

            with open(
                filename, "w", encoding="utf-8"
            ) as file:

                file.write("=" * 55 + "\n")
                file.write("             EMPLOYEE PAYSLIP\n")
                file.write("=" * 55 + "\n")

                file.write(
                    "Generated on: "
                    + datetime.now().strftime("%Y-%m-%d %H:%M")
                    + "\n\n"
                )

                file.write(
                    f"Employee ID : {employee.employee_id}\n"
                )

                file.write(
                    f"Name        : {employee.name}\n"
                )

                file.write(
                    f"Designation : {employee.designation}\n\n"
                )

                file.write("EARNINGS\n")
                file.write("-" * 55 + "\n")

                file.write(
                    f"Basic Salary : Rs. {employee.basic_salary:.2f}\n"
                )

                file.write(
                    f"Allowances   : Rs. {employee.allowances:.2f}\n"
                )

                file.write(
                    f"Gross Salary : Rs. {employee.gross_salary():.2f}\n\n"
                )

                file.write("DEDUCTIONS\n")
                file.write("-" * 55 + "\n")

                file.write(
                    f"Total Deductions: Rs. "
                    f"{employee.deductions:.2f}\n\n"
                )

                file.write("-" * 55 + "\n")

                file.write(
                    f"NET SALARY: Rs. {employee.net_salary():.2f}\n"
                )

                file.write("=" * 55 + "\n")

            print("Payslip generated successfully!")
            print("Saved as:", filename)

            # Display the payslip in the terminal too

            employee.display()

        except OSError as error:

            print("Error generating payslip:", error)

    # ------------------------------------------------------
    # UPDATE EMPLOYEE
    # ------------------------------------------------------

    def update_employee(self, employee_id):

        employee = self.find_employee(employee_id)

        if employee is None:

            print("Employee not found.")
            return

        print("Enter the updated employee details.")

        name = get_text("New name: ")
        designation = get_text("New designation: ")

        basic_salary = get_money("New basic salary: ")
        allowances = get_money("New allowances: ")
        deductions = get_money("New deductions: ")

        if deductions > basic_salary + allowances:

            print("Deductions cannot exceed gross salary.")
            return

        updated_employee = Employee(
            employee_id,
            name,
            designation,
            basic_salary,
            allowances,
            deductions
        )

        index = self.employees.index(employee)
        self.employees[index] = updated_employee

        print("Employee record updated successfully!")

        self.save_csv()

    # ------------------------------------------------------
    # SAVE EMPLOYEE DATA TO CSV
    # ------------------------------------------------------

    def save_csv(self):

        try:

            with open(
                "employees.csv", "w",
                newline="", encoding="utf-8"
            ) as file:

                writer = csv.writer(file)

                writer.writerow([
                    "Employee ID",
                    "Name",
                    "Designation",
                    "Basic Salary",
                    "Allowances",
                    "Deductions"
                ])

                for employee in self.employees:

                    writer.writerow([
                        employee.employee_id,
                        employee.name,
                        employee.designation,
                        str(employee.basic_salary),
                        str(employee.allowances),
                        str(employee.deductions)
                    ])

            print("Employee records saved to employees.csv.")

        except OSError as error:

            print("Error saving CSV:", error)


    def save_txt(self):

        try:

            with open(
                "payroll_records.txt", "w", encoding="utf-8"
            ) as file:

                file.write("EMPLOYEE PAYROLL RECORDS\n")
                file.write("=" * 65 + "\n\n")

                for employee in self.employees:

                    file.write(
                        f"Employee ID: {employee.employee_id}\n"
                    )

                    file.write(
                        f"Name: {employee.name}\n"
                    )

                    file.write(
                        f"Designation: {employee.designation}\n"
                    )

                    file.write(
                        f"Basic Salary: Rs. "
                        f"{employee.basic_salary:.2f}\n"
                    )

                    file.write(
                        f"Allowances: Rs. "
                        f"{employee.allowances:.2f}\n"
                    )

                    file.write(
                        f"Gross Salary: Rs. "
                        f"{employee.gross_salary():.2f}\n"
                    )

                    file.write(
                        f"Deductions: Rs. "
                        f"{employee.deductions:.2f}\n"
                    )

                    file.write(
                        f"Net Salary: Rs. "
                        f"{employee.net_salary():.2f}\n"
                    )

                    file.write("-" * 65 + "\n")

            print("Payroll records saved to payroll_records.txt.")

        except OSError as error:

            print("Error saving TXT:", error)



    def load_csv(self):

        if not os.path.exists("employees.csv"):

            print("No existing employee records found.")
            return

        try:

            loaded_employees = []

            with open(
                "employees.csv", "r",
                newline="", encoding="utf-8"
            ) as file:

                reader = csv.DictReader(file)

                for row in reader:

                    basic_salary = Decimal(row["Basic Salary"])
                    allowances = Decimal(row["Allowances"])
                    deductions = Decimal(row["Deductions"])

                    if (
                        basic_salary < 0
                        or allowances < 0
                        or deductions < 0
                        or deductions > basic_salary + allowances
                    ):
                        raise ValueError(
                            "Invalid salary data in CSV."
                        )

                    employee = Employee(
                        row["Employee ID"],
                        row["Name"],
                        row["Designation"],
                        basic_salary,
                        allowances,
                        deductions
                    )

                    if any(
                        item.employee_id == employee.employee_id
                        for item in loaded_employees
                    ):
                        raise ValueError(
                            "Duplicate employee ID in CSV."
                        )

                    loaded_employees.append(employee)

            self.employees = loaded_employees

            print("Employee records loaded successfully.")

        except (OSError, ValueError, KeyError) as error:

            print("Could not load employee records:", error)


# ==========================================================
# INPUT HELPER FUNCTIONS
# ==========================================================

def get_text(prompt):

    while True:

        value = input(prompt).strip()

        if value:
            return value

        print("This field cannot be empty.")


def get_money(prompt):

    while True:

        try:

            amount = Decimal(input(prompt).strip())

            if not amount.is_finite():
                print("Enter a valid monetary amount.")
                continue

            if amount < 0:
                print("Amount cannot be negative.")
                continue

            if amount != amount.quantize(Decimal("0.01")):
                print("Enter no more than two decimal places.")
                continue

            return amount

        except InvalidOperation:

            print("Please enter a valid number.")



# MAIN PROGRAM


payroll = EmployeePayrollSystem()


while True:


    print("       EMPLOYEE PAYROLL SYSTEM")
    print("=" * 45)

    print("1. Add Employee")
    print("2. Calculate Salary")
    print("3. Generate Payslip")
    print("4. Search Employee")
    print("5. Display All Employees")
    print("6. Update Employee")
    print("7. Save Payroll Data to CSV")
    print("8. Save Payroll Data to TXT")
    print("9. Save Data to Both Files")
    print("10. Exit")



    choice = input("Enter your choice: ").strip()

    try:

        if choice == "1":

            print("\nADD EMPLOYEE")

            employee_id = get_text("Employee ID: ")
            name = get_text("Employee Name: ")
            designation = get_text("Designation: ")

            basic_salary = get_money("Basic Salary (Rs.): ")
            allowances = get_money("Allowances (Rs.): ")
            deductions = get_money("Deductions (Rs.): ")

            if deductions > basic_salary + allowances:

                print("Deductions cannot exceed gross salary.")
                continue

            employee = Employee(
                employee_id,
                name,
                designation,
                basic_salary,
                allowances,
                deductions
            )

            payroll.add_employee(employee)

        elif choice == "2":

            employee_id = get_text("Enter Employee ID: ")

            payroll.calculate_salary(employee_id)

        elif choice == "3":

            employee_id = get_text("Enter Employee ID: ")

            payroll.generate_payslip(employee_id)

        elif choice == "4":

            employee_id = get_text("Enter Employee ID: ")

            payroll.search_employee(employee_id)

        elif choice == "5":

            payroll.display_employees()

        elif choice == "6":

            employee_id = get_text("Enter Employee ID: ")

            payroll.update_employee(employee_id)

        elif choice == "7":

            payroll.save_csv()

        elif choice == "8":

            payroll.save_txt()

        elif choice == "9":

            payroll.save_csv()
            payroll.save_txt()

        elif choice == "10":

            payroll.save_csv()
            payroll.save_txt()

            print("Payroll data saved. Goodbye!")
            break

        else:

            print("Invalid choice. Enter a number from 1 to 10.")

    except (ValueError, InvalidOperation) as error:

        print("Invalid input:", error)

    except OSError as error:

        print("File operation failed:", error)
```
